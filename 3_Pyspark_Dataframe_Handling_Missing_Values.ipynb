## Pyspark Handling Missing Values
- Dropping Rows based on missing values
- Various Parameter In Dropping functionalities
- Fill/Handle Missing values by Mean, Median And Mode

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("MissingValues").getOrCreate()
```

```python
spark
```

```python
df_pyspark = spark.read.csv('test2.csv', header = True)  # for missing valaues
```

```python
df_pyspark.printSchema()
```

```python
df_pyspark = spark.read.csv('test2.csv', header = True, inferSchema=True) # for replacing with mean
```

```python
df_pyspark.printSchema()
```

```python
df_pyspark.show()
```

## 1. Dropping Rows based on missing values

```python
df_pyspark.na.drop().show()
```

## 2. Various Parameter In Dropping functionalities

```python
## how == any -> if row has even a single null value
## how == all -> Only if row has all null values
df_pyspark.na.drop(how='any').show()
```

```python
df_pyspark.na.drop(how='all').show()
```

```python
## thresh == int -> Min. NON NULL values required in row
df_pyspark.na.drop(how = 'any', thresh=2).show()
```

```python
# subset -> Remove rows based on null value in a SPECIFIC COLUMN
df_pyspark.na.drop(how = 'any', subset=['Name']).show()
```

## 3. Fill/ Missing values by Mean, Median And Mode


```python
# Based on all columns
df_pyspark.na.fill('N/A').show()
```

```python
# Based on Specific columns
df_pyspark.na.fill('N/A', ['Experience']).show()
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols=['age', 'Experience', 'Salary'],
    outputCols=["{}_imputed".format(c) for c in ['age', 'Experience', 'Salary']]
).setStrategy("median")
```

```python
imputer.fit(df_pyspark).transform(df_pyspark).show()
```
