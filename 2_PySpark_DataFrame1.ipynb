```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("DataFrame").getOrCreate()
```

```python
spark
```

### 1. Read the Dataset
> `inferSchema=True`: Maintains datatype

```python
df_pyspark = spark.read.option('header', 'true').csv('test1.csv', inferSchema=True)
```

```python
df_pyspark.show()
```

### 2. Checking the Datatypes of the Column(Schema)

```python
df_pyspark.printSchema()
```

```python
df_pyspark = df_pyspark.drop('Salary')
```

```python
df_pyspark.printSchema()
```

### Better reading

```python
df_pyspark = spark.read.csv('test1.csv', header=True, inferSchema=True)
```

```python
df_pyspark.show()
```

```python
df_pyspark.printSchema()
```

### Dropping column

```python
df_pyspark = df_pyspark.drop('Salary')
df_pyspark.printSchema()
```

```python
type(df_pyspark)
```

### 3. Selecting Columns And Indexing

```python
# See columns
df_pyspark.columns
```

```python
df_pyspark.head(3)
```

```python
# Select specific column
df_pyspark.select('Name')
```

```python
df_pyspark.select('Name', 'Experience').show()
```

### Check data type

```python
# Check data type
df_pyspark.dtypes
```

## 4. Check Describe option similar to Pandas

```python
# Describe
df_pyspark.describe()
```

```python
df_pyspark.describe().show()
```

## 5. Adding Columns

```python
df_pyspark = df_pyspark.withColumn('Experience after 2 years', df_pyspark['Experience']+2)
```

```python
df_pyspark.show()
```

## 6. Dropping Columns

```python
df_pyspark = df_pyspark.drop('Experience after 2 years')
```

```python
df_pyspark.show()
```

## 7. Rename column

```python
df_pyspark.withColumnRenamed('Name', 'First Name').show()
```
